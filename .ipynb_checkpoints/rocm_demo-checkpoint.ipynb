# ROCm GPU Demo: System Information

```python

import subprocess

print("=== ROCm Information ===")
result = subprocess.run(['rocminfo'], capture_output=True, text=True)
print(result.stdout)

print("\n=== System Information ===")
result = subprocess.run(['lspci', '-v'], capture_output=True, text=True)
print(result.stdout[:1000])
```

```python

import matplotlib.pyplot as plt
import numpy as np

# GPU memory data
labels = ['Total', 'Used']
sizes = [34.2, 17.0]
colors = ['#1f77b4', '#ff7f0e']

plt.figure(figsize=(10, 6))
bars = plt.bar(labels, sizes, color=colors, edgecolor='black')

plt.ylabel('Memory (GB)', fontsize=12)
plt.title('GPU Memory Usage', fontsize=14, fontweight='bold')
plt.ylim(0, 40)

# Add values on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 1,
             f'{height}GB', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.savefig('/tmp/gpu_memory_chart.png', dpi=300, bbox_inches='tight')
print("✅ Chart saved to /tmp/gpu_memory_chart.png")
print("📊 Memory data: Total=34.2GB, Used=17.0GB")
```
